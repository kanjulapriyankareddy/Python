```python
def calculate_area(length,width):
    """
    this function calculates area of rectangle
    input: length and width
    output: area of rectangle
    created date: 8-9-2026
    """
    return length*width
```

```python
def calculate_perimeter(length,width):
    """
    this function calculates perimeter of rectangle
    input: length and width
    output: perimeter of rectangle
    created date: 8-9-2026
    """
    return 2*(length+width)
```

```python
def print_shape(length,width):
    """
    print the area and perimeter of a rectangle.

    parameters:
    length(float): length of rectangle
    width(float): width of rectangle
    created date: 8-9-2026
    """
    area = calculate_area(length, width)
    perimeter = calculate_perimeter(length, width)
    print(f"rectangle with length {length} and width {width}:")
    print(f"area of rectangle is {area}")
    print(f"perimeter of rectangle is {perimeter}")
```

```python
print_shape(5, 10)
```

```python
print_shape(15,10)
```

# Decarator


A Python decorator is a design pattern that allows you to modify or extend the behavior of a function or method without directly changing its source code

```python
def deco(func):
  def fun():
    print("*************")
    func()
    print("##############")
  return fun   
```

```python
@deco
def hello():
    print("hello people")
```

```python
hello()
```

```python
@deco
def hello2():
    print("bhavana joined the class")
```

```python
hello2()
```

```python
def doctor_only(func):
    def wrapper(role,patient_name):
        if role=="doctor":
            return func(role,patient_name)
        else:
            print("access denied")
    return wrapper        
```

```python
@doctor_only
def view_patient(role,patient_name):
    print(f"patient record displayed for: {patient_name}")
```

```python
view_patient("doctor","raju")
```

```python
view_patient("nurse","raju")
```

```python
def add_consultation_fee(func):
    def wrapper(patient,doctor):
        print(f"patient:{patient}")
        print(f"Doctor: {doctor}")
        print("consultation fee: 500 INR")
        func(patient,doctor)

    return wrapper    
```

```python
@add_consultation_fee
def appointment(patient,doctor):
    print("Appointment booked successfully")
```

```python
appointment("raju","Dr.Bhavana")
```

- else:

the else block will be executed if no exceptions is raised in the try block

if exception block is raised in the try block, else block will not be  excuted

```python
try:
 f=open("files/sample2.txt","r")
except Exception as e:
   print(e)

else:
    print(f.read())  
```

#custom exception


```python
# define a custom exception
class validate_age(Exception):
    def __init__(self,message="Age is not valid"):
        self.message=message
        super().__init__(self.message)
# use custom exception in function        
def validate_age1(age):
        if age<0:
            raise validate_age("age should be positive")
        elif age>200:
            raise validate_age("age must be lessthan 200")
        else:
            print("age is valid")

     #handle the exception
try:
    age=int(input("enter the age:"))
    validate_age1(age)

except validate_age as e: 
    print(e)          
```
