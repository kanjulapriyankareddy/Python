BREAK,CONTINUE,PASS

Break:

- when specific condition met, break allows to terminate the loop

```python
a="bhavana" 
for i in a:
   if i=="v":
      break
   print(f"this is break statement:{i}")      
```

continue:

when specific condition met, continue allows to skip the current execution of the loop and control  is transfered to the starting of the loop.

```python
a="bhavana"
for i in a:
    if i=="v":
        continue
    print(f"this is continue statement:{i}")
```
