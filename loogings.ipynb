Logging:

it is a crucial aspect of any application

it is used to track the events,errors, total operational information

pyhton built-in logging module offers a flexible framework for emiting log messages from python program


```python
import logging
import os
```

```python
log_dir= "loggings"
log_file_name="app.log"

os.makedirs(log_dir, exist_ok=True)
log_path=os.path.join(log_dir,log_file_name)
```

```python
logging.basicConfig(filename=log_path, level=logging.INFO)
```

LOG  are 5 types

INFO:

CONFIRMATION THAT THINGS ARE WORKING  AS EXPECTED

```python
logging.info("This is an info message that will be logged to the file.")
```

warning:

An indication that something unexpected happend or indicating of some problem in the near future

example:

low diskspace


```python
logging.warning("this is first warning")
```

Critical:

it is very serious problem ,it effect the continous running of the problem

```python
logging.critical("this is my critical error")
```

error:

due to some serious problem, that software has not been able to perform some function

```python
logging.error("this is my first error")
```

Debug:

detailed information, it works when only diagnosing the problem


```python
logging.debug("this is my firdt debug")
```

```python
log_dir= "loggings"
log_file_name="app.log"

os.makedirs(log_dir, exist_ok=True)
log_path=os.path.join(log_dir,log_file_name)
```

```python
logging.basicConfig(filename=log_path,level=logging.INFO,format="[%(asctime)s] %(name)s- %(levelname)s-%(message)s",force=True)
```

```python
def add(a,b):
    result=a+b
    logging.info(f"adding {a}+{b}={result}")
    return result
```

```python
add(10,20)
```

```python
def sub(a,b):
    result= a-b
    logging.info(f"subtraction {a}-{b}={result}")
    return result

sub(10,20)  
```

```python
def multiply(a,b):
    result=a*b
    logging.info(f"multiplicatin {a} *{b}={result}")
    return result    
multiply(10,20)
```

```python
def divide(a,b):
    try:
        result=a/b
        logging.debug(f"dividing {a}/{b}={result}")
        return result
    except ZeroDivisionError:
        logging.error("division by zero error")
        return None
divide(10,0)    
```
