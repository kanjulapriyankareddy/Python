# static method

- static method can be called without self or cls

-this method can access any attribute like either class attribute or instance attribute

- the decorator written as @static method


```python
class mymath:
    @staticmethod
    def add(x,y):
        return x+y

mymath.add(10,20)    
```

```python
class netflix:
    new_show="kalki"
    def __init__(self,mail,password):
        self.mail=mail
        self.password=password
        self.service="free"
        self.amount=0

    def get_password(self):
        return self.password

    @staticmethod
    def get_newshow():
        return netflix.new_show
    
    @staticmethod
    def get_mail(a):
        return a.mail
```

# encapsulation

- it is a pillar of oops

- using encapsulated variables amd methods we can controll access to the data

- a core concept in programming where data and the functions that use that data are bundled together into a single unit

# public members 

  it can access anywhere(inside the class or outside of the class)

  self.name=name

# protected member:
- the protected members start with one underscore

- it can be accessed by the class and its subclasses but not outside of the class    
    
-    self._name=name

# private member

- private members start with double underscore.

- the private member can't access directly

- to access private member we have to (getter) method and for modification use (setter) method

-      self.__name=name

```python
class mentor:
    def __init__(self,name):
        self.name=name

obj=mentor("bhavana")   # here you can pass the   any value of your choice to the constructor and it will be assigned to the instance attribute name 
print(obj.name)       
```

```python
class mentor:
    def __init__(self):
        self.name="bhavana"

obj=mentor()  # here it assigns the value "bhavana" to the instance attribute name when the object is created   it is not necessary to pass any value to the constructor since it is already assigned to the instance attribute name
print(obj.name)        
```

```python
class mentor:
    def __init__(self):
        self.name="bhavana"
obj=mentor("priyanka")
print(obj.name)        
```

# protected member

```python
# protected member
class mentor:
  def __init__(self):
    self._name='bhavana'

obj=mentor()
print(obj._name)  # accessing the protected member using instance   

      
```

# protected member with subclasses

```python
class father:
    def __init__(self):
        self._salary=10000  

class son(father):
    def display_salary(self):
        print(self._salary)  # accessing the protected member from the parent class     
obj=son()
obj.display_salary()        
```

# for accessing of private member we have to use getter method

```python
class mentor:
    def __init__(self):
     self.__name="bhavana"

    def get_name(self):
        return self.__name  

obj=mentor()
print(obj.get_name())  # accessing the private member using instance method     
        
```

# for modifying private data using setter method

```python
class mentor:
    def __init__(self):
        self.__name="bhavana"
    def set_mentorname(self,change):
        self.__name=change  # modifying the private member using instance method
    def get_mentorname(self):
        return self.__name  # accessing the private member using instance method       
obj=mentor()
obj.set_mentorname("priyanka")  # modifying the private member using instance method
```

```python
print(obj.get_mentorname())  # accessing the private member using instance method
```
